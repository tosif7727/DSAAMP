```python
# Import required modules
from selenium import webdriver           # Selenium controls web browsers
from selenium.webdriver.chrome.options import Options # To set browser options
from selenium.webdriver.common.by import By           # For finding elements
import time                              # To manage delays
import pandas as pd                      # For storing data in a DataFrame

# Ethical scraping: set options for headless browsing (not opening a window)
chrome_options = Options()
chrome_options.add_argument("--headless")  # Run browser in the background

# Ethical scraping: identify yourself with a custom user-agent
chrome_options.add_argument(
    "user-agent=Mozilla/5.0 (compatible; MyWebScraper/1.0; +http://yourdomain.com/bot)"
)

# Ethical scraping: don't scrape forbidden pages, so we check robots.txt first
import requests
robots_url = "https://quotes.toscrape.com/robots.txt"
robots_txt = requests.get(robots_url).text
print("robots.txt content:\n", robots_txt)
# Here, you should read and follow the rules!

# Start the browser
driver = webdriver.Chrome(options=chrome_options)

# URL of the dynamic site for practice
url = "https://quotes.toscrape.com/js/"
driver.get(url)  # Open the web page

# Wait for JavaScript to load content (ethical: avoid hammering the server)
time.sleep(3)  # You can use WebDriverWait for more advanced waits

# Extract data (quotes and authors)
quotes = []
authors = []
tags = []  # Optional: to store tags if needed
h1 = []  # Optional: to store h1 text if needed
h1 = driver.find_element(By.TAG_NAME, "h1").text  # Optional: get h1 text (removed, not present in .quote)
# h1.append(h1)  # Store h1 text if needed
# Ethical: scrape only the visible, public data
quote_elements = driver.find_elements(By.CLASS_NAME, "quote")
for quote in quote_elements:
    # Each quote has text and author
    text = quote.find_element(By.CLASS_NAME, "text").text
    author = quote.find_element(By.CLASS_NAME, "author").text
    tag = quote.find_element(By.CLASS_NAME, "tag").text  # Optional: get tags
    quotes.append(text)
    authors.append(author)
    tags.append(tag)  # Optional: store tags if needed

# Store the data in a DataFrame (for easy use later)
df = pd.DataFrame({
    "Quote": quotes,
    "Author": authors,
    "Tag": tags,  # Optional: include tags if needed
    "H1": h1  # Optional: include h1 text if needed

})

# Save data ethically for your own study, not for resale
df.to_csv("quotes_scraped.csv", index=False)

# Always close the browser to free resources
driver.quit()

print(df.head())
```
